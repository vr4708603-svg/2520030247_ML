```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv("/content/placementfilledcat.csv")

# Select only two numerical features
X = df[['CodingTestScore', 'AptitudeTestScore']]

# Apply Standard Scaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert to DataFrame
scaled_df = pd.DataFrame(X_scaled, columns=['CodingTestScore', 'AptitudeTestScore'])

print("Original Data")
print(X.head())

print("\nAfter Standard Scaling")
print(scaled_df.head())
```
